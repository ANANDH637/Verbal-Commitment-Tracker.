```python

# Verbal Commitment Tracker

import re

commitment_patterns = [
    r"\bI will\b",
    r"\bI'll\b",
    r"\bI am responsible for\b",
    r"\bI can handle\b",
    r"\bI promise to\b"
]

deadline_pattern = (
    r"\b(?:by|before|on)\s+"
    r"(Monday|Tuesday|Wednesday|Thursday|Friday|Saturday|Sunday|"
    r"tomorrow|today|next week|next month)\b"
)

def track_commitments(transcript):
    sentences = re.split(r'(?<=[.!?])\s+', transcript.strip())
    commitments = []

    for sentence in sentences:
        matched = any(
            re.search(pattern, sentence, re.IGNORECASE)
            for pattern in commitment_patterns
        )

        if matched:
            deadline_match = re.search(
                deadline_pattern, sentence, re.IGNORECASE
            )

            deadline = (
                deadline_match.group(1)
                if deadline_match else "Not specified"
            )

            # Extract the speaker's stated task after the commitment phrase
            task = re.sub(
                r"^(.*?\b(?:I will|I'll|I am responsible for|"
                r"I can handle|I promise to)\b)\s*",
                "",
                sentence,
                flags=re.IGNORECASE
            ).strip(" .")

            commitments.append({
                "statement": sentence,
                "task": task,
                "deadline": deadline
            })

    print("=" * 50)
    print("VERBAL COMMITMENT TRACKER")
    print("=" * 50)

    if not commitments:
        print("No predefined commitment phrases found.")

    for i, item in enumerate(commitments, start=1):
        print(f"\nCommitment {i}")
        print("Statement:", item["statement"])
        print("Task:", item["task"])
        print("Deadline:", item["deadline"])

    print("\nTotal commitments detected:", len(commitments))
    print("=" * 50)


# Sample meeting transcript
meeting = (
    "I will complete the Python module by Friday. "
    "I'll prepare the project report by tomorrow. "
    "We discussed the design during the meeting. "
    "I am responsible for testing by Monday."
)

track_commitments(meeting)
```
